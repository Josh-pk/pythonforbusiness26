```python
#Task 1 A list of dictionaries

transactions = [
    {"id": 1, "product": "Laptop", "price": 1200, "quantity": 2, "region": "EU"},
    {"id": 2, "product": "Phone", "price": 800, "quantity": -1, "region": "US"},
    {"id": 3, "product": "Tablet", "price": 0, "quantity": 5, "region": "EU"},
    {"id": 4, "product": "Monitor", "price": 350, "quantity": 3, "region": "US"},
    {"id": 5, "product": "Laptop", "price": 1200, "quantity": 1, "region": "EU"},
    {"id": 6, "product": "Keyboard", "price": 75, "quantity": 10, "region": "APAC"},
]

# a) Filter invalid transactions
valid_transactions = (
    t for t in transactions #"t for t in" assings t as an item in "transactions", which calls for the items only.
    if t["price"] > 0 and t["quantity"] >= 0
)

# b) Computing new field 
for t in valid_transactions:
    t["total"] = t["price"] * t["quantity"]

# c) Group by region and compute total revenue
revenue_by_region = {}

for t in valid_transactions:
    region = t["region"]
    reveneu_by_region[region] = revenue_by_region.get(region, 0) + t["total"]

# d) Find the region with the highest revenue
max_region = max(revenue_by_region)
max_revenue = reveneu_by_region[max_region]

print(f"The region with the highest total revenue is {max_region} with {max_revenue}.")
```

```python
#Task 2 Creating a function

sales = [1200, 980, 1540, 870, 2010, 1650, 1300, 1100, 950, 1780, 2200, 1430, 1600, 1900]

def summarise_sales(sales_list):
    total = sum(sales_list)
    average = round(total / len(sales_list), 2)
    best_day = sales_list.index(max(sales_list)) + 1
    worst_day = sales_list.index(min(sales_list)) + 1
    above_average_count = sum(1 for sale in sales_list if sale > average)

    return {
    "total": total,
    "average": average,
    "best_day": best_day,
    "worst_day": worst_day,
    "above_average_count": above_average_count
    }

# Call the function with the list above and print the results
summary = summarise_sales(sales)
print(summary)
```
