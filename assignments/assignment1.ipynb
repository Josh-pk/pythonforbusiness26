```python
#Task 1 A list of dictionaries

transactions = [
    {"id": 1, "product": "Laptop", "price": 1200, "quantity": 2, "region": "EU"},
    {"id": 2, "product": "Phone", "price": 800, "quantity": -1, "region": "US"},
    {"id": 3, "product": "Tablet", "price": 0, "quantity": 5, "region": "EU"},
    {"id": 4, "product": "Monitor", "price": 350, "quantity": 3, "region": "US"},
    {"id": 5, "product": "Laptop", "price": 1200, "quantity": 1, "region": "EU"},
    {"id": 6, "product": "Keyboard", "price": 75, "quantity": 10, "region": "APAC"},
]

# a) Filter invalid transactions
valid_transactions = []
for t in transactions: #"t for t in" assings t as an item in "transactions", which calls for the items only.
    if t["price"] > 0 and t["quantity"] >= 0:
        valid_transactions.append(t)


# b) Computing new field 
for t in valid_transactions:
    t["total"] = t["price"] * t["quantity"]
    

# c) Group by region and compute total revenue
revenue_by_region = {}

for t in valid_transactions:
    region = t["region"]
    if region not in revenue_by_region:
        revenue_by_region[region] = 0
    revenue_by_region[region] += t["total"]

# d) Find the region with the highest revenue
best_region = 0
highest = 0

for region in revenue_by_region:
    if highest is None or revenue_by_region[region] > highest:
        highest = revenue_by_region[region]
        best_region = region

print(f"The region with the highest total revenue is {best_region} with revenue of {highest}.")
```

```python
#Task 2 Creating a function

sales = [1200, 980, 1540, 870, 2010, 1650, 1300, 1100, 950, 1780, 2200, 1430, 1600, 1900]

def summarise_sales(sales_list):
    total = sum(sales_list)
    average = round(total / len(sales_list), 2)
    best_day = sales_list.index(max(sales_list)) + 1
    worst_day = sales_list.index(min(sales_list)) + 1
    above_average_count = sum(1 for sale in sales_list if sale > average)

    return {
    "total": total,
    "average": average,
    "best_day": best_day,
    "worst_day": worst_day,
    "above_average_count": above_average_count
    }

# Call the function with the list above and print the results
summary = summarise_sales(sales)
print(summary)
```

```python
#Task 3 Loops
students = [
    {"name": "Anna", "grade": 85, "attendance": 0.92},
    {"name": "Erik", "grade": 58, "attendance": 0.75},
    {"name": "Mia", "grade": 91, "attendance": 0.98},
    {"name": "Leo", "grade": 44, "attendance": 0.60},
    {"name": "Sara", "grade": 72, "attendance": 0.88},
    {"name": "Tom", "grade": 67, "attendance": 0.70},
    {"name": "Nina", "grade": 95, "attendance": 0.99},
    {"name": "Olli", "grade": 50, "attendance": 0.65},
]

# a) Create a list of those who passed
passing = []
for s in students:
    if s["grade"] >= 50:
        passing.append(s["name"])

print(passing)

# b) Loop through the students and print a status message for each. If the student's grade is 80 or above
for s in students:
    if s["grade"] >= 80:
        print(s["name"] + ": Excellent")
    elif s["grade"] >= 50:
        print(s["name"] + ": Pass")
    else:
        print(s["name"] + ": Fail")

# c) Count attendance rate of below 0.80
count = 0
for s in students:
    if s["attendance"] < 0.80:
        count += 1

print(count)

# d) Print the student with highest grade
top = students[0]
for s in students:
    if s["grade"] > top["grade"]:
        top = s

print(top["name"], top["grade"])

# e) Print student with highest grade
top_1 = max(students, key=lambda s: s["grade"])
print(top_1["name"], top_1["grade"])
```

```python
#Task 4 Class
class Portfolio:
    def __init__(self, owner):
        self.owner = owner
        self.holdings = {}  # ticker -> {shares, price_per_share}

# a) Adding assets
    def add_asset(self, ticker, shares, price_per_share):
        self.holdings[ticker] = {
            "shares": shares,
            "price_per_share": price_per_share
        }

# b) Total market value
    def total_value(self):
        total = 0
        for ticker in self.holdings:
            shares = self.holdings[ticker]["shares"]
            price = self.holdings[ticker]["price_per_share"]
            total += shares * price
        return total

# c) Return the ticker of the asset with the highest total value
    def most_valuable(self):
        best = 0
        highest = 0

        for ticker in self.holdings:
            value = self.holdings[ticker]["shares"] * self.holdings[ticker]["price_per_share"]
            if value > highest:
                highest = value
                best = ticker

        return best

# d) Return a readable summary of the portfolio, then print three instances
    def __str__(self):
        return f"{self.owner}'s portfolio has {len(self.holdings)} assets worth {self.total_value()}"

p = Portfolio("Batman")
p.add_asset("BAT", 10, 200)
p.add_asset("ROBN", 5, 250)
p.add_asset("JKR", 2, 3000)

print(p)
print(p.total_value())
print(p.most_valuable())
```
